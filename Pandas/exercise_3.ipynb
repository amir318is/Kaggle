```python
def count_tropical(row):
    k =0
    if 'tropical' in row.description:
        k = k+1 
    return k

def count_fruity(row):
    k =0
    if 'fruity' in row.description:
        k = k+1 
    return k
descriptor_counts = pd.Series([reviews.apply(count_tropical,axis='columns').value_counts()[1], reviews.apply(count_fruity,axis='columns').value_counts()[1]],index=['tropical', 'fruity'])
descriptor_counts
```

```python
def star_rating(row):
    if row.country == 'Canada':
        return 3
    elif 95 <= row.points:
        return 3
    elif 85 <= row.points < 95:
        return 3
    else:
        return 1

star_ratings = pd.Series(reviews.apply(star_rating, axis='columns'))
star_ratings
```
